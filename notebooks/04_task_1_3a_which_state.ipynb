# Task 1.3(a): what state did we actually make?

**In plain words.** Every process except single-photon loss moves photons in **pairs**, so whether the photon count is even or odd (the *parity*) never changes. The memory starts empty (even), so it must end in the **even cat** $|C_+\rangle = (|\alpha\rangle+|-\alpha\rangle)/\mathcal N$, a superposition of both valleys, not one valley.

**How we tell a superposition from a 50/50 mixture:**

| | parity ⟨P⟩ | purity Tr ρ² | coherence C | Wigner fringes |
|---|---|---|---|---|
| even cat | +1 | 1 | 1 | yes |
| mixture | 0 | 0.5 | 0 | no |

Here $C = 2\,|\langle\alpha|\rho_a|-\alpha\rangle|$, the size of the off-diagonal term between the two valleys.

Single-photon loss flips parity, so the cat should fade into a mixture at rate $\Gamma \approx 2\kappa_a\bar n$ (about 0.51 per µs, lifetime ≈ 1.9 µs).
