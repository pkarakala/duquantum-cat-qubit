# Task 1.3(a): identify the prepared state and its decay

Inflation from vacuum prepares a partially decohered even cat. Without memory single-photon loss, stabilization approaches the even superposition $\lvert C_+\rangle\propto\lvert\alpha\rangle+\lvert-\alpha\rangle$: two-photon exchange preserves memory parity, and vacuum is even. Memory single-photon loss flips parity and makes the state mixed during inflation.

Here $\rho_a$ is the density matrix after tracing out the buffer. We track parity $\langle\Pi\rangle$, with $\Pi=(-1)^{a^\dagger a}$; even-cat fidelity $F_+$; purity $\mathcal{P}$; and inter-lobe coherence $C$. At 1 µs, $F_+=0.811$, $\mathcal{P}=0.692$ and $C=0.626$: interference remains substantial, although the state is already partially mixed. An equal incoherent mixture has $C\approx0.00030$.

Fitting parity over 1–6 µs gives decay rate $\Gamma=0.5135$ µs⁻¹, close to the photon-loss estimate $2\kappa_a\langle n_a\rangle_{\mathrm{ss}}=0.5146$ µs⁻¹. Its inverse, 1.95 µs, is the time for an approximately $e$-fold reduction after inflation. By 6 µs the diagnostics approach those of the incoherent mixture.
